### Passo a passo de análise de dados em python

- Passo 2: Visualizar e tratar essa base de dados
- Passo 3: "Dar uma olhada" na sua base de dados
- Passo 4: Construir uma análise para identificar o motivo de cancelamento
  - Identificar qual o motivo ou os principais motivos dos clientes estarem cancelando o cartão de crédito.

```python
#importando as bibliotecas necessárias
import pandas as pd
```

```python
#Fazendo o ETL
#1. Extração
#impotando a base de dados.
tabela = pd.read_csv("ClientesBanco.csv", encoding="latin1")
tabela = tabela.drop("CLIENTNUM", axis=1)
display(tabela)
```

```python
#2. Transformação
tabela = tabela.dropna()
display(tabela.info())
display(tabela.describe().round(1))
```

### agora vavos fazer uma EDA - Análise Exploratoria dos dados.
 - vamos avaliar como esta a divisão entre os Cliente x Cancelados.

```python
qtde_categoria = tabela["Categoria"].value_counts()
display(qtde_categoria)

qtde_categoria_perc = tabela["Categoria"].value_counts(normalize=True).round(1)*100
display(qtde_categoria_perc)
```

### Temos vários formas de descobrir o motivo de cancelamento
  - Podemos olhar a comparação entre Clientes e Cancelados em cada uma das colunas da nossa base de dados, para ver se essa informação traz algum insight novo para a gente

```python
import plotly.express as px
import plotly.io as pio

pio.renderers.default = "notebook" #ou "browser" para abrir no navegador,"vscode".
for coluna in tabela:
  grafico = px.histogram(tabela, x=coluna, color="Categoria")
  grafico.show("png")
  
```

### Informações retiradas da análise

- Me parece que quanto mais produtos contratados um cliente tem, menor a chance dele cancelar.
- E quanto mais transações e quanto maior o valor de transação, menor a chance dele cancelar.
- Quanto maior a quantidade de contatos que a pessoa teve que fazer, maior a chance dela cancelar.
#### Perfil dos clientes que cancelam os cartões
- Cerca de 80% dos clientes usam o plano Blue. 
- 90% dos clientes estão entre a faixa de 30 à 60 anos.
- 30% dos clientes possuem ensino médio, e 30% ensino superior, 30% não possuem ensino formal ou não especificaram.
- cerca de 35% dos clientes ganham menos de 40k por ano.
- cerca de 24% dos clientes possuem vínculo como cliente em pelo menos 36 menses.
- cerca de 80% dos cliente passam até 3 meses sem usarem o cartão ou inativos.
